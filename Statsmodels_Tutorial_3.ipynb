```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
df = sm.datasets.get_rdataset("Guerry","HistData").data
df = df[["Lottery","Literacy","Wealth","Region"]].dropna()
res = smf.ols(formula="Lottery ~ Literacy + Wealth + Region", data=df).fit()
res.summary()
```

```python
#Use C(...) to expicitly mention a variable to be categorical 
res_c = smf.ols(formula="Lottery ~ Literacy + Wealth + C(Region)", data=df).fit()
res_c.summary()
```

```python
#This is to neglect the intercept
res_nc = smf.ols(formula="Lottery ~ Literacy + Wealth + C(Region) - 1", data=df).fit()
res_nc.summary()
```

INTERACTION EFFECT

```python
#Interaction: Matrix, this will only show the interaction term's p-value
res_int1 = smf.ols(formula="Lottery ~ Literacy:Wealth - 1", data=df).fit()
res_int1.params
```

```python
#Interaction: individual columns, this will show the p-value for the individual columns as well as teh interaction term
res_int2 = smf.ols(formula="Lottery ~ Literacy*Wealth - 1", data=df).fit()
print(res_int2.params)
```

FUNCTION USAGE WHILE INITIALIZING THE OLS

```python
#overwriting the res model
res = smf.ols(formula="Lottery ~ np.log(Literacy)", data=df).fit()
res.summary()
```

```python
#custom function
def log_plus_1(x):
    return np.log(x) + 1
```

```python
res = smf.ols(formula="Lottery ~ log_plus_1(Literacy)", data=df).fit()
res.summary()
```
