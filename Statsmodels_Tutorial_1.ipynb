```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

```python
#grabs data and return pandas dataframe object
data = sm.datasets.get_rdataset("Guerry","HistData").data
data.head()
```

```python
data.columns
```

```python
#R-style formula
model_ols = smf.ols("Lottery ~ Literacy + np.log(Pop1831)", data=data)
res = model_ols.fit()
res.summary()
```

THE FIRST LINE IS TO INITIALIZE THE BLUEPRINT THAT WE'RE CREATING WITH THE OLS APPROACH. SO THAT WHEN WE FIT THE MODEL USING ".fit()" IT USES THE OLS METHOD TO ESTIMATE THE MODEL PARAMETERS.
THE ".summary()" FUNCTION ONLY WORKS ON "RegressionWrapper"(s). FOR THAT REASON WE'VE KEPT THE OLS BLUEPRINT AND THE FINAL "RegressionWrapper" OBJECT SEPARATE

```python
#using numpy arays to do the same thing - (not really looking fwd to using this)
nobs = 100
X = np.random.random((nobs,2))
X = sm.add_constant(X)
```

```python
beta = [1,.1,.5] # coeff
e = np.random.random(nobs) #epsilon error
y = np.dot(X,beta) + e #the final model, finding the dot product
```

```python
model_numpy_ols = sm.OLS(y,X)
res2 = model_numpy_ols.fit()
res2.summary()
```
